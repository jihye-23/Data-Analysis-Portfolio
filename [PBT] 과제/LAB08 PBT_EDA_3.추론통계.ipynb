## #02. 상관분석

### 1. (연속형 종속변수 X 연속형 독립변수)

```python
for f in feature_continuous:
    result = my_stats.correlation(df, x=f, y=target, plot=False)
    print(f"\n독립변수 :{f}")
    display(result)
```

<인사이트>

- 검정 선택 : 모든 쌍에서 정규성 위배(nomality = False) -> Spearman p 자동 적용(비모수) 12개 변수 모두 p < 0.001로 유의

- 채택 기준(PDF) : p < 0.05 & |r| >= 0.3 -> 독립변수 후보로 채택

- 핵심: LSTAT (하위 계층 비율)이ㅣ |p|=0.85로 MEDV와 가장 강한 음(-)의 상관, RM(방 개수)이 가장 강한 양(+)의 상관 -> 두 변수가 집값 설명력의 핵심 후보
- B만 |p|<0.3(약함)이라 유일하게 후보에서 제외, 나머지 11개는 채택

=> 상관관계다 라는 것이 인과관계라는 것은 아님 : 강한 p라도 원인으로 단정지으면 안된다.

## #03. T검정

### 1. 이진 명목형 변수 X 연속형

- CHAS (강 인접)에 따른 MEDV 차이
    - 검정값을 wide 포맷 (두 컬럼)으로 분리해 전달

```python
chas_medv = DataFrame({
    "비인접(0)" : df.loc[df["CHAS"] == 0, target].reset_index(drop=True),
    "인접(1)": df.loc[df["CHAS"] == 1, target].reset_index(drop=True)
})

my_stats.test_independent(chas_medv, "비인접(0)", "인접(1)", title="CHAS에 따른 MEDV 차이")
```

<인사이트>

명목(2) X 연속 : T검정 계열 , 정규성 위배 -> Mann-Whitney U(비모수) 자동 적용

- 해석 : 찰스강 인접 여부에 따라 집값이 통계적으로 유의하게 다르며, 인접 지역이 더 비쌈
- p < 0.05 유의 -> CHAS를 독립변수 후보로 채택
- 비모수 검정이라 Cohen's d가 출력되지 않음 -> 실질 크기는 박스플롯의 중앙값 차이로 보완 판단

## #04. ANOVA

### 1. 3개 이상의 그룹을 갖는 명목형 파생변수 생성

- 원본엔 3집단 + 명목형이 없기 때문에 가상의 파생변수를 생성해서 진행한다
- 이 검정의 결과는 최종 변수 선별에는 포함되지 않는다.

```python
df["RM_grp"] = cut(df["RM"], 3, labels=["방적음", "방보통", "방많음"])
df["RM_grp"].value_counts().sort_index()
```

### 2. 일원분산분석

```python
my_stats.anova_oneway(df, y="MEDV", between="RM_grp")
```

<인사이트>
- 명목(3+) X 연속 -> ANOVA.RM을 3구간(방적음/방보통/방많음)으로 나눈 파생변수로 검정. 등분산 위배 -> Welh ANOVA 자동 적용

- 해석 : 방개수 구간에 따라 MEDV가 통계적으로 유의하게 다르며, 효과크기도 매우 큼
- F만으로는 "어느 쌍이" 다른지 알 수 없음 -> 사후검정 필요(다음 셀)
- RM_grp는 실습용 파생변수이기 때문에 최종 변수 선별에는 포함하지 않는다.

### 2. 사후검정

```python
my_stats.posthoc_oneway(df, y="MEDV", between="RM_grp", title="RM 구간별 MEDV 사후검정")
```

<인사이트>

정규성. 등분산 위배 -> 사후검정은 Games-Howell적용. 세 쌍 모두 유의

- 방향 : 방적음(15.3) < 방보통(20.7) < 방많음(38.6) -> 방이 많을수록 집값이 뚜렷하게 상승
- 모든 쌍이 유의, 특히 방많음 그룹이 다른 두 그룹과 매우 큰(Large) 차이를 보임
- 방 개수(RM)가 집값을 단조 증가시키는 강한 요인임을 사후검정이 재확인
